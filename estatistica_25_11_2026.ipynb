```python
import pandas as pd

df_produtos = pd.read_csv("/content/produtos.csv")
df_vendas = pd.read_csv("/content/vendas.csv")
```

```python
display(df_produtos)
df_produtos.info()
df_produtos.describe(include= "all")
```

resumo estatistico

```python
#aplicar resumo estatistico

#queremos saber a media gasta nos produtos vendidos(preço médio das vendas)

# somar o valor total das vendas e dividir pelo total
df_produtos_vendas =pd.merge(df_vendas, df_produtos, on='id_produto')

df_produtos_vendas['valor_da_venda'] = df_produtos_vendas['quantidade'] * df_produtos_vendas['preco_unitario']
display(df_produtos_vendas)

media_vendas = df_produtos_vendas['valor_da_venda'].mean()
print(media_vendas)

moda_produto = df_produtos_vendas['produto'].mode()[0]
print(moda_produto)

```

```python
moda_linha =df_produtos_vendas['linha'].mode()
print(moda_linha)
```
